# Syncing
There are a few ways that I want to be able to sync

* Sync from the front
    This is the simplest type, if a video is a straight rip from Twitch to YT vod, then the zero point will both be the same
* Sync from the end
    This is also very simple, sometimes streamers will play copyrighted music at the start of stream, and when they upload the vod to YT, they will cut the beginning. If the end of the stream is the same as the one on YT, then the sync point is at the end of the video.
* Sync from the middle
    This one is a bit tricky, sometimes the vod won't sync up with the front or end of the video, this can happen when a stream gets cut into multiple days of a subathon. Usually the twitch stream will be one giant file, but the YT can only be 12 or 24 hours long. Thus, you will have to find a sync point manually.


# Sync from the middle
Most streams display chat throughout the stream. Given this you should be able to find a sync point from something unique from the comments.
* A unique message that was sent "THIS SONG IS AMAZING! MELODY IS SUCH A GOOD EARWORM!!!" (easiest)
* Order of usernames ("harder")

# Find the times by the Order of usernames

```python
# unique order is 2,3,2
order = [1,2,3,4,5,1,2,3,2,3,2,4,5,1,2,5]

# find when the unique order happens from chat
unique_order = [2,3,2]

def find_subarray(arr, subarr):
    for i in range(len(arr) - len(subarr) + 1):
        if arr[i:i + len(subarr)] == subarr:
            return i
    return -1

def find_subarray_occurrences(arr, subarr):
    times_found = []
    for i in range(len(arr) - len(subarr) + 1):
        if arr[i:i + len(subarr)] == subarr:
            times_found.append(i)
    return times_found

find_subarray(order, unique_order)
find_subarray_occurrences(order, unique_order)
```
